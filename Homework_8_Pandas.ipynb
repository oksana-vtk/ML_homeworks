**Про датасет:**

Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

**Колонки датасету:**
- `total_bill` - загальна сума рахунку (долари)
- `tip` - розмір чайових (долари)
- `sex` - стать відвідувача (Male/Female)
- `smoker` - чи курить (Yes/No)
- `day` - день тижня (Thur/Fri/Sat/Sun)
- `time` - час відвідування (Lunch/Dinner)
- `size` - розмір групи (кількість людей)

**Джерело:** Оригінальні дані зібрані в одному з американських ресторанів у 1990-х роках. Скачати і переглянути набір даних можна тут: https://www.kaggle.com/datasets/sakshisatre/tips-dataset

**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
average_bill = df['total_bill'].mean()
print(f"Average bill: ${average_bill:.2f}")
print(f"Max tips: ${df['tip'].max()}")

print(f"\nTotal missing values in all columns:\n{df.isnull().sum()}\n")
print(f"Total missing values in dataframe: {df.isnull().sum().sum()}\n")

lunch_and_dinner = df[(df["time"] == "Lunch") | (df["time"] == "Dinner")]
print(f"Total number of bills for Lunch and Dinner: {len(lunch_and_dinner)}")

number_of_lunch_bills = df['time'].value_counts().get('Lunch', 0)
number_of_dinner_bills = df['time'].value_counts().get('Dinner', 0)
number_of_lunch_and_dinner_bills = number_of_lunch_bills + number_of_dinner_bills

print(f"\nNumber of Lunch bills: {number_of_lunch_bills}")
print(f"Number of Dinner bills: {number_of_dinner_bills}")
print(f"Number of Lunch and Dinner bills: {number_of_lunch_and_dinner_bills}\n")
```

---

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python

avg_bill_by_sex = df.groupby('sex')["total_bill"].mean()
print(f"Average bill by sex:\n{avg_bill_by_sex.round(2)}\n")

avg_tips_by_day = df.groupby("day")["tip"].mean()
print(f"Average tips by day:\n{avg_tips_by_day.round(2)}\n")

cliens_per_day = df.groupby("day")["size"].sum()
print(f"Client per day:\n{cliens_per_day}\n")

bills_per_day = df.groupby("day")["total_bill"].count()
print(f"Bills per day:\n{bills_per_day.round(2)}\n")

```

**6. Створення нової колонки**

Створіть нову колонку `tip_percentage` - відсоток чайових від загальної суми рахунку.
Виведіть перші 10 рядків з новою колонкою.

```python
df["tip_percentage"] = (df["tip"] / df["total_bill"] * 100).round(2)
print(df.head(10))
```

**7. Сортування**

Відсортуйте датасет і виведіть результат сортування в кожному випадку окремо (кожен пункт оцінюється в 0.5 бала):
- За розміром рахунку (зростання)
- За чайовими (спадання)

Додатково виведіть топ-5 найбільших чайових використовуючи метод `nlargest`.

```python
df_sorted_bills = df.sort_values("total_bill", ascending=True)
print(f"Sorted by Total bill ASC:\n{df_sorted_bills.head(10)}\n")

df_sorted_tips = df.sort_values("tip", ascending=False)
print(f"Sorted by Tip DESC:\n{df_sorted_tips.head(10)}\n")

# Display the top 5 largest tips
print(df.nlargest(5, "tip"))
```
