**1. Створення змінних**

Створіть змінні для особистої інформації (кожен пункт оцінюється в 0.5 бала):

**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var_1 = 15
var_2 = 4

result_1 = var_1 + var_2
result_2 = var_1 - var_2
result_3 = var_1 * var_2
result_4 = var_1 / var_2
result_5 = var_1 // var_2
result_6 = var_1 % var_2
result_7 = var_1 ** var_2

print(f"Сума {var_1} і {var_2}: {result_1}")
print(f"Різниця {var_1} і {var_2}: {result_2}")
print(f"Добуток {var_1} і {var_2}: {result_3}")
print(f"Ділення {var_1} і {var_2}: {result_4}")
print(f"Цілочисельне ділення {var_1} і {var_2}: {result_5}")
print(f"Залишок від ділення {var_1} і {var_2}: {result_6}")
print(f"Піднесення до степеня {var_1} і {var_2}: {result_7}")
```
