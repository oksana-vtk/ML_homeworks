**1. Цикл for - базовий**

Використовуючи цикл `for`, виведіть числа від 1 до 7 включно використовуючи функцію range.


```python
for i in range(8):
  print(i)
```


---

**2. Цикл for по списку**

Дано список:

```python
names = ["Олег", "Анна", "Іван", "Марія"]
```


Використовуючи цикл, виведіть кожне ім'я з привітанням: "Привіт, [ім'я]!"


```python
for name in names:
  print(f"Привіт, {name}!")
```


---

**3. Цикл з enumerate**

Дано список оцінок:

```python
grades = [85, 92, 78, 95, 88]
```


Використовуючи `enumerate`, виведіть кожну оцінку з номером: "Студент 1: 85 балів"


```python
for i, grade in enumerate(grades):
  print(f"Студент {i + 1}: {grade} балів")
```

**4. Сума парних чисел**

Напишіть програму, яка обчислює суму всіх парних чисел від 1 до 20 включно.

```python
# За домогою for та if:
total_sum = 0
for i in range(21):
  if i % 2 == 0:
    total_sum += i
print(f"1. Сума парних чисел від 1 до 20 включно: {total_sum}")

# За допомогою List Comprehansion та функції sum:
total_sum_1 = sum([i for i in range(21) if i % 2 ==0])
print(f"2. Сума парних чисел від 1 до 20 включно: {total_sum_1}")

# За допомогою функції sum та range
total_sum_2 = sum(range(2,21,2))
print((f"3. Сума парних чисел від 1 до 20 включно: {total_sum_2}"))
```
